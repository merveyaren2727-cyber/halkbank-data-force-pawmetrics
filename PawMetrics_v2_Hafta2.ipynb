```python
# ==============================================================================
# PAWMETRICS v2.0 - AI KARAR DESTEK VE PROMPT MÜHENDİSLİĞİ MODÜLÜ (HAFTA 2)
# Halkbank Data Force | 17 & 19 Kasım - AI Okuryazarlığı I & II
# ==============================================================================

# 1. SYSTEM INSTRUCTION (ROL + BAĞLAM + GÖREV + HALÜSİNASYON KISITI)
SYSTEM_INSTRUCTION = """
[ROL]: Sen Evcil Hayvan Bakım ve Bütçe Optimizasyon Asistanısın.
[BAĞLAM]: Sana yalnızca PawMetrics kural motorundan gelen stok (gün) ve bütçe sapması (TL) verileri verilecektir.
[GÖREV]: Her ürün için sayısal veriye dayalı, tek cümlelik net bir aksiyon önerisi üret.
[KISITLAR - HALÜSİNASYON KONTROLÜ]:
1. Sana verilen sayısal değerler dışında kesinlikle marka, piyasa fiyatı veya uydurma oran ekleme.
2. Yorumunu yalnızca 'kalan_gun' ve 'butce_farki' değişkenlerine dayandır.
3. Çıktıyı standart karar formatında sun.
"""

# 2. 1. HAFTA İZ TABLOSUNDAN GELEN DOĞRULANMIŞ VERİLER (CONTEXT INJECTION)
iz_tablosu_verileri = [
    {"urun": "Kısırlaştırılmış Somonlu Mama (15kg)", "kalan_gun": 52, "butce_farki": +250},
    {"urun": "Bentonit Kedi Kumu (20L)", "kalan_gun": 30, "butce_farki": -70},
    {"urun": "Ödül & Vitamin", "kalan_gun": 20, "butce_farki": +20}
]

# 3. HALÜSİNASYON KONTROLLÜ AI KARAR SİMÜLASYONU
def ai_karar_motoru(veri, kritik_esik=15):
    urun = veri["urun"]
    gun = veri["kalan_gun"]
    sapma = veri["butce_farki"]

    # Bağlam (Context) tabanlı deterministik-AI hibrit muhakeme
    if gun <= kritik_esik and sapma > 0:
        return f"ACİL SİPARİŞ & BÜTÇE REVİZYONU: Stok {gun} güne düştü ve +{sapma} TL aşım var; toplu alım indirimi araştırılmalı."
    elif gun > kritik_esik and sapma > 0:
        return f"STOK GÜVENLİ / MALİYET UYARISI: {gun} günlük stok yeterli; +{sapma} TL sapmayı dengelemek için sonraki alımda kampanya beklenmeli."
    elif gun <= kritik_esik and sapma <= 0:
        return f"SİPARİŞ ZAMANI: Bütçe hedefte ({sapma} TL) ancak stok {gun} güne indi; planlı sipariş oluşturulmalı."
    else:
        return f"OPTİMAL DURUM: {gun} günlük stok yeterli ve bütçede {abs(sapma)} TL tasarruf sağlandı; mevcut rutin korunmalı."

# 4. ÇIKTI RAPORU
print("=== 1. YAPILANDIRILMIŞ SİSTEM TALİMATI (SYSTEM PROMPT) ===")
print(SYSTEM_INSTRUCTION.strip())
print("\n" + "="*85)
print("=== 2. PAWMETRICS v2.0 HALÜSİNASYON KONTROLLÜ AI KARAR ÇIKTISI ===")
print("="*85)

for kayit in iz_tablosu_verileri:
    oneri = ai_karar_motoru(kayit)
    print(f"📦 ÜRÜN  : {kayit['urun']} (Stok: {kayit['kalan_gun']} gün | Sapma: {kayit['butce_farki']:+d} TL)")
    print(f"🤖 AI    : {oneri}")
    print("-" * 85)
```
