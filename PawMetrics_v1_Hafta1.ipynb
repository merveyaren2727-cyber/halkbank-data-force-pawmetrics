```python
# PawMetrics v1.0 - Akıllı Stok Tahmini ve Bütçe Sapma Motoru

kalemler = ["Kısırlaştırılmış Somonlu Mama (15kg)", "Bentonit Kedi Kumu (20L)", "Ödül & Vitamin"]
mevcut_stok_gram = [4500, 12000, 300]       # Kalan miktar (gram/ml)
gunluk_tuketim = [85, 400, 15]             # Günlük ortalama tüketim
gerceklesen_harcama = [2450, 680, 420]     # Aylık gerçekleşen maliyet (TL)
planlanan_butce = [2200, 750, 400]         # Aylık hedef bütçe (TL)

kritik_gun_esigi = 15

print("=== PAWMETRICS v1.0 HAFTALIK DURUM RAPORU ===\n")

for i in range(len(kalemler)):
    urun = kalemler[i]
    kalan_gun = int(mevcut_stok_gram[i] / gunluk_tuketim[i])
    butce_farki = gerceklesen_harcama[i] - planlanan_butce[i]

    if kalan_gun <= kritik_gun_esigi:
        stok_durumu = "⚠️ KRİTİK STOK - Sipariş Ver!"
    else:
        stok_durumu = "✅ Stok Yeterli"

    if butce_farki > 0:
        butce_durumu = f"🔴 Bütçe Aşımı (+{butce_farki} TL)"
    else:
        butce_durumu = f"🟢 Bütçe İçi ({butce_farki} TL)"

    print(f"Ürün: {urun}")
    print(f"  -> Tahmini Ömür: {kalan_gun} gün | Durum: {stok_durumu}")
    print(f"  -> Bütçe Analizi: {butce_durumu}\n")
```
